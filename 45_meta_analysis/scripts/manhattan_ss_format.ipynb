```r
library(dplyr)
library(qqman)
```

## Read Data

```r
ham_meta <- read.table("../output/meta_results1.tbl", header = T)
```

## Format Data

```r
#begin by transforming the p-valu
head(ham_meta)
ham_meta$log10_p_val <- -log10(ham_meta$P.value)
```

```r
 colnames(ham_meta) <- c("SNP", "A1","A2","FREQ_A1", "FREQ_A1_SE",
                        "MIN_FREQ_A1", "MAX_FREQ_A1", "BETA", "SE",
                         "P", "DIRECTION", "LOG10_P")
```

```r
# Split the SNP name
snp_split <- strsplit(ham_meta$SNP, split = ':')

# Save chromosome and position
ham_meta$CHR <- unlist(lapply(snp_split, function(x){x[1]})) |> as.numeric()
ham_meta$BP <- unlist(lapply(snp_split, function(x){x[2]})) |> as.numeric()
```

```r
#unlist(lapply(t, function(x){x[1]}))
head(ham_meta)
```

## Make Plots


```r
png("../output/meta_ham_manhattan.png")
manhattan(ham_meta, ylim = c(0, 8), 
    col = c("BROWN3", "DEEPSKYBLUE3","DARKOLIVEGREEN3","MEDIUMPURPLE"))
dev.off()
```

```r
png("../output/ham_qq_regenie.png")
qq(ham_meta$P,xlim = c(0, 8), ylim = c(0, 8))
dev.off()
```

## Format Sum Stats to Save


```r
write.table(ham_meta,"../output/meta_ham_ss.txt", row.names = F, col.names = T, quote = F)
```
